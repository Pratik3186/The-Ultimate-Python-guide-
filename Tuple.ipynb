```python
def min_in_tuple(tpl):
  return min(tpl)
def max_in_tpl(tpl):
  return max(tpl)
def sum_of_tpl(tpl):
  return sum(tpl)

tpl = (1,2,3,4,5)
print(min_in_tuple(tpl))
print(max_in_tpl(tpl))
print(sum_of_tpl(tpl))
```
