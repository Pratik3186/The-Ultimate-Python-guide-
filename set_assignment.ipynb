```python
a = {1,2,3,4,5}
b = {6,8,10,12,14}
print(f"Union: {a.union(b)}")
print(f"Intersection: {a.intersection(b)}")
print(f"Diffrence: {a.difference(b)}")
print(f"Symmetric diffrence: {a.symmetric_difference(b)}")
```
